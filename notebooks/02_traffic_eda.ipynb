```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import plotly.express as px
import seaborn as sns
```

```python
df = pd.read_csv("../data/clean_traffic.csv")
df
```

```python
junction_average = df.groupby("Junction")["Vehicles"].mean()
print(junction_average)
```

# Average Traffic Per Junction

```python
junction_average.plot(kind = "bar", figsize = (5,3), color = "skyblue")
plt.xlabel("Junction")
plt.ylabel("Average Number of Vehicles")
plt.title("Average Traffic Volume at Each Junction")
plt.xticks(rotation = 0)
plt.show()
```

# Average Traffic Per Hour

```python
hourly_average = df.groupby("Hour")["Vehicles"].mean()
plt.plot(hourly_average.index, hourly_average.values)
plt.xlabel("Hour of the Day")
plt.ylabel("Average Number of Vehicles")
plt.title("Average Traffic Volume by Hour of the Day")
plt.xticks(rotation = 0)
plt.grid()
plt.show()
```

# Hourly Traffic Pattern

```python
hourly_traffic = df.groupby('Hour')['Vehicles'].mean()

plt.figure(figsize=(12,5))

plt.plot(hourly_traffic.index, hourly_traffic.values)

plt.title("Average Traffic by Hour")
plt.xlabel("Hour")
plt.ylabel("Average Vehicles")

plt.grid(True)

plt.show()
```

# Traffic Distribution

```python
plt.figure(figsize=(10,5))

plt.hist(df['Vehicles'], bins=30)

plt.title("Vehicle Distribution")
plt.xlabel("Vehicles")
plt.ylabel("Frequency")

plt.show()
```

# Traffic Trend Over Time

```python
daily_traffic = df.groupby('Day')['Vehicles'].mean()

plt.figure(figsize=(10,5))

plt.plot(daily_traffic.index, daily_traffic.values)

plt.title("Average Daily Traffic")
plt.xlabel("Day")
plt.ylabel("Average Vehicles")

plt.grid(True)

plt.show()
```

```python
numeric_df = df[['Vehicles', 'Hour', 'Day', 'Month', 'Weekday']]
numeric_df.corr()
```

# Correlation Analysis

```python
plt.figure(figsize=(8,5))
sns.heatmap(numeric_df.corr(), annot=True)
plt.title("Correlation Heatmap")
plt.show()
```
